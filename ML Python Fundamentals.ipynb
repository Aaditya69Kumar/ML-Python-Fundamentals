Dataset: A collection of data, usually in a table format (like an Excel or CSV file), that we use to teach the machine.

Features (X): These are the Inputs or "Clues". They are the independent variables the model uses to make a prediction (e.g., age, cholesterol, or blood pressure).

Target (y): This is the Output or the "Answer". It is the specific thing we want the machine to predict (e.g., whether a patient has heart disease or not).

Training: This is the phase where the machine looks at the data (X and y) to find patterns. In your code, model.fit(X, y) is the training step.

Model: Often called the "Brain," this is the mathematical algorithm (like Linear Regression) that learns from the data.

Prediction: Once the model is trained, we give it new, unseen data (X) and ask it to guess the result (y). In your code, this is model.predict().

Supervised Learning: Like a student with a teacher. We give the machine the questions (X) and the correct answers (y) so it can learn.

Regression: Predicting a continuous number (e.g., predicting a house price or a specific age).

Classification: Predicting a category or "Label" (e.g., Pass vs. Fail, or Healthy vs. Sick).

Pandas: The "Data Librarian." It is used to load and organize data into a table called a DataFrame (df).

Scikit-Learn (sklearn): The "Machine Learning Toolkit." It contains all the pre-built models like Linear Regression.

Matplotlib: The "Artist." It is used to create charts and graphs so we can visualize our data.

In traditional coding, we write the rules. In Machine Learning, we give the computer Features (X) and a Target (y), and it learns the rules itself

```python
# @title
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Data: x is input (Features), y is output (Target) -> Rule is y = 2x
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# 2. Initialize the Model (The Brain)
model = LinearRegression()

# 3. Training: The machine looks for the pattern
model.fit(x, y)

# 4. Prediction: Test it with a new number!
prediction = model.predict([[10]])
print(f"For input 10, the machine learned the answer is: {prediction[0]}")
```

# First Machine Learning Model (Linear Regression)
